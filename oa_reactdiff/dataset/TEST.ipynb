```python
import torch

# Define periods (rows) and groups (columns) for main elements
# Periods 1-7, Groups 1-18, skip lanthanoids and actinoids
period_group = {
    1: (1, 1),  2: (1, 18),                   # H, He
    3: (2, 1),  4: (2, 2),  5: (2, 13),  6: (2, 14),  7: (2, 15),  8: (2, 16),  9: (2, 17),  10: (2, 18), # Li-Ne
    11: (3, 1), 12: (3, 2), 13: (3, 13), 14: (3, 14), 15: (3, 15), 16: (3, 16), 17: (3, 17), 18: (3, 18),
    19: (4, 1), 20: (4, 2), 21: (4, 3), 22: (4, 4), 23: (4, 5), 24: (4, 6), 25: (4, 7), 26: (4, 8), 27: (4, 9), 28: (4, 10),
    29: (4, 11), 30: (4, 12), 31: (4, 13), 32: (4, 14), 33: (4, 15), 34: (4, 16), 35: (4, 17), 36: (4, 18),
    37: (5, 1), 38: (5, 2), 39: (5, 3), 40: (5, 4), 41: (5, 5), 42: (5, 6), 43: (5, 7), 44: (5, 8), 45: (5, 9), 46: (5, 10),
    47: (5, 11), 48: (5, 12), 49: (5, 13), 50: (5, 14), 51: (5, 15), 52: (5, 16), 53: (5, 17), 54: (5, 18),
    55: (6, 1), 56: (6, 2), # skip 57-71 lanthanoids
    72: (6, 4), 73: (6, 5), 74: (6, 6), 75: (6, 7), 76: (6, 8), 77: (6, 9), 78: (6, 10),
    79: (6, 11), 80: (6, 12), 81: (6, 13), 82: (6, 14), 83: (6, 15), 84: (6, 16), 85: (6, 17), 86: (6, 18),
    87: (7, 1), 88: (7, 2), # skip 89-103 actinoids
    104: (7, 4), 105: (7, 5), 106: (7, 6), 107: (7, 7), 108: (7, 8), 109: (7, 9), 110: (7, 10),
    111: (7, 11), 112: (7, 12), 113: (7, 13), 114: (7, 14), 115: (7, 15), 116: (7, 16), 117: (7, 17), 118: (7, 18)
}

# Generate one-hot vectors
row_one_hot = {}
col_one_hot = {}

for z, (row, col) in period_group.items():
    row_vec = [0 for _ in range(7)]
    col_vec = [0 for _ in range(18)]
    row_vec[row-1] = 1.0
    col_vec[col-1] = 1.0
    row_one_hot[z] = row_vec
    col_one_hot[z] = col_vec

def get_periodic_one_hot(atomic_number):
    """Returns the one-hot encoded period and group vectors for a given atomic number."""
    return row_one_hot[atomic_number],col_one_hot[atomic_number]
    
# Example usage:
z = 8  # Oxygen
print("Row one-hot:", row_one_hot[z])
print("Column one-hot:", col_one_hot[z])

get_periodic_one_hot(z)
```
